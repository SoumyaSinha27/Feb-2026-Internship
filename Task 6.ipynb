```python
#Question: 01
def parking_system(capacity, logs):
    parked = 0
    peak = 0
    
    for log in logs:
        if log == "IN":
            parked += 1
        elif log == "OUT" and parked > 0:
            parked -= 1
        
        # Track peak usage
        if parked > peak:
            peak = parked
    
    print("Currently Parked Vehicles:", parked)
    
    if parked > capacity:
        print("Parking Status: Full - Capacity Exceeded")
    elif parked == capacity:
        print("Parking Status: Full")
    else:
        print("Parking Status: Available")
    
    print("Peak Usage:", peak)

parking_system(50, ["IN", "IN", "IN", "OUT", "IN", "IN", "OUT"])
```

```python
#Question: 02
def estimate_delivery(distance, traffic, weather):
    base_time = distance * 5 # 5 minutes per km
    
    # Traffic adjustment
    if traffic == "High":
        base_time += 15
    elif traffic == "Medium":
        base_time += 10
    else:
        base_time += 5
    
    # Weather adjustment
    if weather == "Rainy":
        base_time += 10
    elif weather == "Storm":
        base_time += 20
    
    print("Estimated Delivery Time:", base_time, "minutes")

estimate_delivery(8, "High", "Rainy")
```

```python
#Question: 03
def theatre_occupancy(total_seats, booked_seats):
    booked_count = len(booked_seats)
    
    occupancy = (booked_count / total_seats) * 100
    
    print("Occupancy:", int(occupancy), "%")
    
    if occupancy == 100:
        print("Show Status: Housefull")
    elif occupancy >= 75:
        print("Show Status: Almost Full")
    else:
        print("Show Status: Available")
    
    if occupancy >= 90:
        print("Suggestion: Open Additional Show")

theatre_occupancy(200, [1]*150)
```

```python
#Question: 04
def server_load(cpu_readings):
    total = 0
    
    for reading in cpu_readings:
        total += reading
    
    average = total / len(cpu_readings)
    
    print("Average CPU Load:", int(average), "%")
    
    if average < 50:
        print("Server Status: Normal")
    elif average <= 80:
        print("Server Status: Warning")
    else:
        print("Server Status: Critical")

server_load([45, 60, 70, 85, 90])
```

```python
#Question: 05
def classroom_monitor(resources):
    overused = []
    
    for resource, hours in resources.items():
        if hours > 8:
            overused.append(resource)
    
    if overused:
        print("Overused Resources:", ", ".join(overused))
        print("Energy Alert: Yes")
    else:
        print("Overused Resources: None")
        print("Energy Alert: No")

usage = {"Projector": 6,"AC": 9,"Lights": 4}

classroom_monitor(usage)
```

```python
#Question: 06
def event_registration(capacity, registrations):
    
    if registrations > capacity:
        confirmed = capacity
        waitlisted = registrations - capacity
        status = "Closed"
    else:
        confirmed = registrations
        waitlisted = 0
        status = "Open"
    
    print("Confirmed Registrations:", confirmed)
    print("Waitlisted Users:", waitlisted)
    print("Registration Status:", status)

event_registration(100, 105)
```
