```python
# Question :1
def analyze_eng(likes):
    total_likes=0
    for like in likes:
        total_likes+=like
    if total_likes>=1000:
        status="Viral Post"
    else:
        status="Normal"
    print("Total Likes: ",total_likes)
    print("Post Status: ",status)
analyze_eng([200,300,250,300])
```

```python
# Question :2
def check_stock(stock):
    if stock<10:
        status="Low Stock Alert"
    else:
        status="Stock Sufficient"
    print("Medicine Stock: ",stock)
    print("Status: ",status)
check_stock(6)
```

```python
# Question :3
def check_rain(rain_list,req_level):
    total=0
    for rain in rain_list:
        total+=rain
    avg=total/len(rain_list)
    if avg>=req_level:
        status="Adequate Rainfall"
    else:
        status="Inadequate Rainfall"
    print("Average Rainfall: ",avg)
    print("Rainfall Status: ",status)
check_rain([70,75,68,80],70)
```

```python
# Question :4
def detect_dup(uname):
    if len(uname)!=len(set(uname)):
        print("Duplicate Found: Yes")
    else:
        print("Duplicate Found: No")
detect_dup(["Ash","Bob","Ash","Ken"])
```

```python
# Question :5
def check_eligibility(age):
    if age>=18:
        status="Eligible"
    else:
        status="Not Eligible"
    print("Patient Age: ",age)
    print("Eligibility Status: ",status)
check_eligibility(21)
```

```python
# Question :6
def premium(prices):
    pre_list=[]
    for price in prices:
        if price>1000:
            pre_list.append(price)
    print("Premium Crops: ",pre_list)
premium([1500,2500,1800,3200])
```

```python
# Question :7
def health(errors):
    if errors==0:
        Status="Healthy"
    elif errors<=5:
        Status="Minor Issues"
    else:
        Status="Critical Issues"
    print("Error Count: ",errors)
    print("System Status: ",Status)
health(7)
```

```python
# Question :8
def transaction(amt):
    limit=50000
    if amt<=limit:
        status="Approved"
    else:
        status="Rejected"
    print("Transaction Amount: ",amt)
    print("Transaction Status: ",status)
transaction(10000)
```

```python
# Question :9
def attendence(attend_list):
    total=len(attend_list)
    attended=0
    for status in attend_list:
        if status==1:
            attended+=1
    per=(attended/total)*100
    if per>=75:
        eligibility="Eligible"
    else:
        eligibility="Not Eligible"
    print("Attendance Percentage: ",per)
    print("Exam Eligibility: ",eligibility)
attendence([1,1,1,0,1,1,1,1]) #Here 1 means Present and 0 means Absent
```
